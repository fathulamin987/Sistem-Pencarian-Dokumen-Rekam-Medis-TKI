```python
import pandas as pd

df = pd.read_csv(
    "mtsamples.csv",
    sep=";",
    engine="python"
)

print(df.shape)
df.head()
```

```python
df = df[
    [
        'description',
        'medical_specialty',
        'sample_name',
        'transcription',
        'keywords'
    ]
]

print(df.shape)
```

```python
df = df.head(500).copy()

print(df.shape)
```

```python
from deep_translator import GoogleTranslator
import time

def split_text(text, max_chars=4500):
    text = str(text)

    chunks = []

    for i in range(0, len(text), max_chars):
        chunks.append(text[i:i+max_chars])

    return chunks


def translate_long_text(text):
    try:
        chunks = split_text(text)

        hasil = []

        for chunk in chunks:

            translated = GoogleTranslator(
                source='en',
                target='id'
            ).translate(chunk)

            hasil.append(translated)

            time.sleep(0.5)

        return " ".join(hasil)

    except Exception as e:
        print("ERROR:", e)
        return ""
```

```python
# 1. Pastikan kembali datanya siap
df_500 = df.head(500).copy()

# 2. List semua kolom yang mau diterjemahkan sekaligus
kolom_translate = [
    'description',
    'medical_specialty',
    'sample_name',
    'transcription',
    'keywords'
]

# 3. Proses loop otomatis untuk ke-5 kolom dengan print real-time per baris
for kolom in kolom_translate:
    print(f"\n⚡ Memulai Translasi Kolom: [{kolom}]")
    hasil_kolom = []

    for i, text in enumerate(df_500[kolom]):
        terjemahan = translate_long_text(text)
        hasil_kolom.append(terjemahan)

        # Cetak progres tiap baris di baris yang sama (menggunakan \r)
        print(f"  🔄 Sedang memproses baris ke-{i+1}/{len(df_500)}...", end='\r')

        # Setiap kelipatan 50 baris atau baris terakhir, buat baris baru sebagai tanda sukses
        if (i + 1) % 50 == 0 or (i + 1) == len(df_500):
            print(f"  ✅ Baris 1 sampai {i+1} pada kolom [{kolom}] SELESAI!")

    # Membuat kolom baru otomatis sesuai nama aslinya ditambah _id
    df_500[f"{kolom}_id"] = hasil_kolom

print("\n🎉 Selesai! Semua kolom sukses diterjemahkan ke versi _id.")
```

```python
# Menampilkan 5 baris pertama untuk kolom asli vs kolom terjemahan
kolom_pengecekan = [
    'description', 'description_id',
    'medical_specialty', 'medical_specialty_id',
    'sample_name', 'sample_name_id',
    'transcription', 'transcription_id',
    'keywords', 'keywords_id'
]

# Menampilkan data
df_500[kolom_pengecekan].head()
```

```python
# Menghitung jumlah data kosong di kolom terjemahan
for kolom in ['description_id', 'medical_specialty_id', 'sample_name_id', 'transcription_id', 'keywords_id']:
    jumlah_kosong = (df_500[kolom] == '').sum()
    print(f"Jumlah data kosong di {kolom}: {jumlah_kosong} baris")
```

```python
# 1. Menyimpan file dengan format khusus (utf-8-sig) agar tulisan rapi saat dibuka di Excel
df_500.to_csv("hasil_terjemahan_5_kolom.csv", index=False, encoding="utf-8-sig")
print("✅ File CSV berhasil dibuat di penyimpanan Colab!")

# 2. Download file langsung ke komputer/laptop kamu
from google.colab import files
files.download("hasil_terjemahan_5_kolom.csv")
```
